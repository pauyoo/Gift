```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

# Ensure X_train and y_train are clean of NaN values in y
train_data = pd.concat([X_train, y_train], axis=1).dropna(subset=[y_train.name])
X_train_clean = train_data[X_train.columns]
y_train_clean = train_data[y_train.name]

# Initialize the model
model = LogisticRegression()

# Fit the model using the clean data
model.fit(X_train_clean, y_train_clean)

print("Model trained successfully!")
```
